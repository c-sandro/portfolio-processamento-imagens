```python
img = cv2.imread('donqui.jpg')
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
altura = img_gray.shape[0]
largura = img_gray.shape[1]

dx = 400
dy = 100
```

```python
img_trans = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):

    x_trans = x + dx
    y_trans = y + dy

    if 0 <= x_trans and x_trans < largura:
      if 0 <= y_trans and y_trans < altura:
        img_trans[y_trans, x_trans] = img_gray[y, x]
```

```python
plt.figure(figsize=(7,5))
plt.imshow(img_trans, cmap='gray')
plt.title('Imagem transladada')
plt.axis('off')
plt.show()
```

```python
img_wrap = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):

    x_wrap = (x + dx) % largura
    y_wrap = (y + dy) % altura

    img_wrap[y_wrap, x_wrap] = img_gray[y, x]
```

```python
plt.figure(figsize=(7,5))
plt.imshow(img_wrap, cmap='gray')
plt.title('Imagem com wraparound')
plt.axis('off')
plt.show()
```

```python
sx = 1.2
sy = 0.8

largura_scale = int(largura * sx)
altura_scale = int(altura * sy)
```

```python
img_scale = np.zeros((altura_scale, largura_scale), dtype=np.uint8)

for x in range(largura_scale):
  for y in range(altura_scale):

    x_scale = int(x / sx)
    y_scale = int(y / sy)

    img_scale[y, x] = img_gray[y_scale, x_scale]
```

```python
plt.figure(figsize=(7,5))
plt.imshow(img_scale, cmap='gray')
plt.title('Imagem escalada')
plt.axis('off')
plt.show()
```

```python
img_mirrorh = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):

    x_mirrorh = largura-1 - x
    y_mirrorh = y

    img_mirrorh[y, x] = img_gray[y_mirrorh, x_mirrorh]
```

```python
img_mirrorv = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):

    x_mirrorv = x
    y_mirrorv = altura-1 - y

    img_mirrorv[y, x] = img_gray[y_mirrorv, x_mirrorv]
```

```python
img_mirrorboth = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):

    x_mirrorboth = largura-1 - x
    y_mirrorboth = altura-1 - y

    img_mirrorboth[y, x] = img_gray[y_mirrorboth, x_mirrorboth]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(21, 5))

axes[0].imshow(img_mirrorh, cmap='gray')
axes[0].set_title("Espelho horizontal")
axes[0].axis('off')

axes[1].imshow(img_mirrorv, cmap='gray')
axes[1].set_title("Espelho vertical")
axes[1].axis('off')

axes[2].imshow(img_mirrorboth, cmap='gray')
axes[2].set_title("Espelho total")
axes[2].axis('off')

plt.show()
```
