```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
img_clara = cv2.imread('imagem_clara.png')
img_clara
```

```python
img_escura = cv2.imread('imagem_escura.jpg')
img_escura
```

```python
img_neutra = cv2.imread('imagem_neutra.png')
img_neutra
```

```python
img_clara_gray = cv2.cvtColor(img_clara, cv2.COLOR_BGR2GRAY)
img_escura_gray = cv2.cvtColor(img_escura, cv2.COLOR_BGR2GRAY)
img_neutra_gray = cv2.cvtColor(img_neutra, cv2.COLOR_BGR2GRAY)
```

```python
img_clara_vetor = img_clara_gray.ravel()
img_escura_vetor = img_escura_gray.ravel()
img_neutra_vetor = img_neutra_gray.ravel()
```

```python
clara_array = np.zeros(256)
for pixel in img_clara_vetor:
  clara_array[pixel] += 1

escura_array = np.zeros(256)
for pixel in img_escura_vetor:
  escura_array[pixel] += 1

neutra_array = np.zeros(256)
for pixel in img_neutra_vetor:
  neutra_array[pixel] += 1
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 4))

axes[0, 0].imshow(img_clara_gray, cmap='gray', vmin=0, vmax=255)
axes[0, 0].set_title("Imagem Clara")
axes[0, 0].axis('off')

axes[1, 0].bar(range(256), clara_array)

axes[0, 1].imshow(img_escura_gray, cmap='gray', vmin=0, vmax=255)
axes[0, 1].set_title("Imagem Escura")
axes[0, 1].axis('off')

axes[1, 1].bar(range(256), escura_array)

axes[0, 2].imshow(img_neutra_gray, cmap='gray', vmin=0, vmax=255)
axes[0, 2].set_title("Imagem Neutra")
axes[0, 2].axis('off')

axes[1, 2].bar(range(256), neutra_array)

plt.savefig('histogramas.png', bbox_inches='tight', dpi=300)
plt.show()
```

Imagem clara: há claros picos de valores, provavelmente equivalentes aos valores de cor usados para um elemento (fundo, banner da classe, etc). além disso, os picos são majoritariamente nos alcances claros, por ser uma print de uma tela branca.

Imagem escura: ela possui uma distrubuição concentrada nos tons escuros, consistente com o fato de que a foto é escura

Imagem neutra: há bastante valores no alcance claro, provavelmente pelo fundo ser de um tom claro, mas há valores equilibrados por todo o histograma.

Será utilizado a expansão de contraste, para pincipalmente ver os detalhes melhor na imagem escura, mas também para ver o que acontece em uma imagem mais neutra e em uma imagem com picos claros e singulares.

Imagem clara: H = 255, L = 86 (mais ou menos o pico de menor valor)

Imagem escura: H = 80 (ponto onde os valores ficam muito pequenos), L = 0

Imagem neutra: H = 230 (ponto onde os valores caem abruptamente), L = 20 (mais ou menos o valor do pico de menor valor)

```python
def expansao_contraste(imagem, L, H):
    if H == L:
        return imagem.copy()

    resultado = (imagem.astype(float) - L) * 255 / (H - L)

    return resultado.astype(np.uint8)
```

```python
img_clara_expandida = expansao_contraste(img_clara_gray, 86, 255)
img_escura_expandida = expansao_contraste(img_escura_gray, 0, 80)
img_neutra_expandida = expansao_contraste(img_neutra_gray, 20, 230)
```

```python
clara_expandida_array = np.zeros(256)
for pixel in img_clara_expandida:
  clara_expandida_array[pixel] += 1

escura_expandida_array = np.zeros(256)
for pixel in img_escura_expandida:
  escura_expandida_array[pixel] += 1

neutra_expandida_array = np.zeros(256)
for pixel in img_neutra_expandida:
  neutra_expandida_array[pixel] += 1
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 4))

axes[0, 0].imshow(img_clara_expandida, cmap='gray', vmin=0, vmax=255)
axes[0, 0].set_title("Imagem Clara")
axes[0, 0].axis('off')

axes[1, 0].bar(range(256), clara_expandida_array)

axes[0, 1].imshow(img_escura_expandida, cmap='gray', vmin=0, vmax=255)
axes[0, 1].set_title("Imagem Escura")
axes[0, 1].axis('off')

axes[1, 1].bar(range(256), escura_expandida_array)

axes[0, 2].imshow(img_neutra_expandida, cmap='gray', vmin=0, vmax=255)
axes[0, 2].set_title("Imagem Neutra")
axes[0, 2].axis('off')

axes[1, 2].bar(range(256), neutra_expandida_array)

plt.savefig('histogramas_expandidos.png', bbox_inches='tight', dpi=300)
plt.show()
```

O que fez menos diferença foi no print do classroom, pois ele é feito de elementos com cores sólidas, ou seja, a mudança de contraste desses valores altera muito pouco ou não altera a silhueta das coisas.
